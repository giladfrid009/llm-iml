```python
from src.inference.vllm_service import *


service = VLLMService(
    gpu_ids=[0],
    model_name="meta-llama/Llama-2-7b-chat-hf",
    host="0.0.0.0",
    port=None,
    startup_timeout=5*60,
)

messeges = [
    [
        {"role": "user", "content": "What is the capital of France?"},
        {"role": "assistant", "content": "The capital of France is Paris."},
    ],
    [
        {"role": "user", "content": "What is the largest mammal?"},
        {"role": "assistant", "content": "The largest mammal is the blue whale."},
    ],
]

texts = [
    "What is the capital of France?",
    "What is the largest mammal?",
]

service.start()
chat_answers = service.chat(messeges)         # batched chat
gen_answers  = service.generate(texts)     # batched generate
service.shutdown()

print(f"Chat Answers:", chat_answers)
print(f"Generate Answers:", gen_answers)
```
